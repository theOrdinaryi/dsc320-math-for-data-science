### Creating a right Riemann sum function

This function calculates a right Riemann sum for a given function over an interval \([a, b]\).  
It splits the interval into `n` equal parts and uses the right endpoint of each subinterval to form rectangles.

I also added basic checks to make sure the inputs make sense before doing the calculation.

```python
def right_riemann_sum(f, a, b, n):
    """
    Compute the right Riemann sum R_n for function f on [a, b] using n rectangles.
    """
        # Make sure n is a valid number of rectangles // Make sure interval is valid
    if not isinstance(n, int) or n <= 0:
        raise ValueError("n will ne a positive integer.")
    if b < a:
        raise ValueError("b will be greater than or equal to a.")
        
        # Width of each rectangle
    dx = (b - a) / n
    total = 0.0

        # Loop through right endpoints // add rectangle areas
    for k in range(1, n + 1):
        
        xk = a + k * dx      # right endpoint
        total += f(xk) * dx # area of rectangle

    return total
```

```python
# test  //  f(x) = x on [0,1], right sum with n=4  // Expected: 0.625
print(right_riemann_sum(lambda x: x, 0, 1, 4))
```

### Defining the function for the problem

This is the function given in the exercise:  
\( f(x) = \sqrt{4 - x^2} \)

This function represents the top right portion of a circle with radius 2 when restricted to the interval \([0, 2]\).

```python
def f(x):
    return math.sqrt(4 - x**2)
```

### Calculating the right Riemann sum with 4 rectangles

Here I use the right Riemann sum function to calculate \( R_4 \) on the interval \([0, 2]\).

```python
# Compute the right Riemann sum // use 4 rectangles
R4 = right_riemann_sum(f, 0, 2, 4)

print("R_4 =", R4)
```

### Sketch for part (a)

For the sketch, I drew the curve \( y=\sqrt{4-x^2} \) on \([0,2]\), which is a quarter-circle of radius 2.

For \(R_4\), I split \([0,2]\) into 4 equal parts (\(\Delta x=0.5\)) and used the right endpoints \(0.5, 1.0, 1.5, 2.0\) for the rectangle heights.

### Using larger values of n

To see how the approximation improves, I calculate the right Riemann sum using larger values of `n`.

As `n` increases, the rectangles get thinner and the approximation should get closer to the actual area under the curve.

```python
# Compute right Riemann sums for increasing values of n
for n in [4, 10, 50, 200, 1000, 5000]:
    approx = right_riemann_sum(f, 0, 2, n)
    print(f"n={n:5d}  R_n={approx:.10f}")
```

### Comparing the results to the exact area

The exact area under this curve is known to be \( \pi \), since it represents a quarter of a circle with radius 2.

I compare the Riemann sum results to `math.pi` to see how close the approximations get.

```python
print("Exact area (pi) =", math.pi)
print("Difference for R_5000 =", abs(right_riemann_sum(f, 0, 2, 5000) - math.pi))
```

### Final takeaway

The results show that as the number of rectangles increases, the right Riemann sums approach \( \pi \).

This confirms that the area under the curve from 0 to 2 is \( \pi \), and the special name is pi because the region is a quarter of a circle with radius 2.
