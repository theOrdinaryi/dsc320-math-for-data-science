```python
import numpy as np
import pandas as pd
```

To estimate the limit of f(x) as x approaches 3, we evaluate the function
at values increasingly close to 3 from both sides.

```python
def f(x):
    return 2*x**3 - 4*x + 1

x_values = [2.9, 2.99, 2.999, 3.001, 3.01, 3.1]
results = pd.DataFrame({
    "x": x_values,
    "f(x)": [f(x) for x in x_values]
})

results
```

As x approaches 3, the function values approach 43.
Therefore, lim_x >> 3, f(x) = 43.

```python
def g(x):
    return (np.exp(x) - 1) / x

x_values = [-0.1, -0.01, -0.001, 0.001, 0.01, 0.1]
results = pd.DataFrame({
    "x": x_values,
    "g(x)": [g(x) for x in x_values]
})

results
```

As x approaches 0, g(x) approaches 1.
Thus, lim_x >> 0, g(x) = 1.

Yes, both limits could be found by direct substitution.
This means both functions are continuous at those points.
Continuity implies the limit equals the function value.

```python
def average_rate_of_change(func, a, b):
    return (func(b) - func(a)) / (b - a)

def example_function(x):
    return 3*x**2

average_rate_of_change(example_function, 2, 5)
```

This function calculates how much a function changes on average
between two input values.

```python
def distance(t):
    return 4.9 * t**2
```

```python
average_rate_of_change(distance, 5, 6)
```

```python
average_rate_of_change(distance, 5, 5.5)
```

```python
average_rate_of_change(distance, 5, 5.1)
```

As the time interval shrinks, the average speed approaches a single value.
This suggests the instantaneous speed at t = 5.

```python
average_rate_of_change(distance, 5, 5.01)
```

```python
def derivative(t):
    return 9.8 * t
```

```python
derivative(5)
```

The derivative value matches the instantaneous speed.
This shows that derivatives represent instantaneous rates of change.
